```python
from langchain_mcp_adapters.client import MultiserverMCPClient
import nest_asyncio
from langchain.agents import create_agent
```

```python
nest_asyncio.apply()

# Conecte-se ao servidor mcp-time para operações que levam em consideração o fuso horário.
# Este servidor baseado em Go fornece ferramentas para hora atual, análise de tempo relativo,
# conversão de fuso horário, aritmética de duração e comparação de tempo.
mcp_client= MultiserverMCPClient(
    {
        "time":{
            "transport": "stdio",
            "command": "npx",
            "args": ["-y", "@theo.foobar/mcp-time"]
        }
    }
)

# Carregar tools do MCP Server
mcp_tools= await mcp_client.get_tools()
print(f"Loaded {len(mcp_tools)} MCP tools: {[t.name for t in mcp_tools]}")
```

```python
agent_with_mcp= create_agent(
    model="openai:gpt-5-mini",
    tools= mcp_tools,
    system_prompt="You are a helpful assistant"
)
```

```python
result= await agent_with_mcp.invoke(
    "messages":[{"rule": "user", "content": "What's the time in SF right now?"}]
)

for msg in result["messages"]:
    msg.pretty_print()
```
