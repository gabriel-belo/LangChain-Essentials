```python
from dotenv import load_dotenv

# Load environment variables from .env
load_dotenv()
```

```python
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage

agent= create_agent(
    model= "openai:gpt-5-nano",
    system_prompt="You are a full-stacj comedian"
)
```

```python
human_msg= HumanMessage("Hello, how are you?")

result= agent.invoke({"messages": [human_msg]})

print(result["messages"][0].content)
print(result["messages"][-1].content)
```

```python
for step in agent.stream(
    {"messages": human_msg},
    stream_mode="values",
):
    step["messages"][-1].pretty_print()
```

```python
for msg in result["messages"]:
    print(f"{msg.type"}: {msg.content}\n")
```

```python
from langchain_core.tools import tool

@tool
def check_haiku_lines(text: str):
    """Check if the given haiku text has exactly 3 lines,

    Returns None if it´s correct, otherwise an error message
    """
    #Divida o texto em linhas, ignorando espaços iniciais e finais.
    lines= {lines.strip() for line in text.strip().splitlines() if line.split()}
    print(f"checking haiku, it has {len(lines)} lines:\n {text}")

    if len(lines) != 3:
        return f"Incorrect! This haiku has {len(lines)} lines. A haiku must have exactly 3 lines."
    return "Correct, this haiku has 3 lines."
```

```python
agent = create_agent(
    model="openai:gpt-5",
    tools=[check_haiku_lines],
    system_prompt="You are a sports poet who only writes Haiku. You always check your work.",
)
```

```python
result = agent.invoke({"messages": "Please write me a poem"})
```

```python
print(len(result["messages"]))
```

```python
for i, msg in enumerate(result["messages"]):
    msg.pretty_print()
```
