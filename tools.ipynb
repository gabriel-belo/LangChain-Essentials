```python
from typing  import Literal
from langchain.tools import tool
from langchain.agents import create_agent
```

```python
@tool
def real_number_calculator(a: float, b: float, operation: Literal["add", "subtract", "multiply", "devide"]) -> float:
    """Perform basic arithmetic operations on two real numbers."""
    print("Invoking calculator tool")

    if operation == "add":
        return a + b
    elif operation == "subtract":
        return a - b
    elif operation == "multiply":
        return a * b
    elif operation == "devide":
        if b == 0:
            raise ValueError("Division by zero is not allowed")
        return a / b
    else:
        raise ValueError(f"Invalid operation: {operation}")
```

```python
agent= create_agent(
    model= "openai:gpt-5-mini",
    tools=[real_number_calculator],
    system_prompt="You are a helpful assistant"
)
```

```python
result= agent.invoke(
    {"messages": [{"role": "user", "content": "what is 3.1234 * 4.1232"}]}
)

print(result["messages"][-1].content)
```

## Adicionando uma descrição mais detalhada
Embora uma descrição básica seja frequentemente suficiente, o LangChain oferece suporte a descrições aprimoradas. O exemplo abaixo usa um método: Descrições de argumentos no estilo do Google. Usado com parse_docstring=True, isso analisará e passará as descrições dos argumentos para o modelo. Você pode renomear a ferramenta e alterar sua descrição. Isso pode ser eficaz quando você estiver compartilhando uma ferramenta padrão, mas desejar instruções específicas para o agente.

```python
@tool(
    "calculator",
    parse_description= True,
    description=(
        "Perform basic arithmetic operationson tow real numbers."
        "Use this whenever you have operations on any numbers, even if they are integers"
    ),
)

def real_number_calculator(a: float, b: float, operation: Literal["add", "subtract", "multiply", "devide"]) -> float:
    """Perform basic arithmetic operations on two real numbers.

    Args:
        a (float): The first number.
        b (float): The second number.
        operation (Literal["add", "subtract", "multiply", "divide"]):
            The arithmetic operation to perform.

            - `"add"`: Returns the sum of `a` and `b`.
            - `"subtract"`: Returns the result of `a - b`.
            - `"multiply"`: Returns the product of `a` and `b`.
            - `"divide"`: Returns the result of `a / b`. Raises an error if `b` is zero.

    Returns:
        float: The numerical result of the specified operation.

    Raises:
        ValueError: If an invalid operation is provided or division by zero is attempted.
    """
    print("🧮  Invoking calculator tool")
    # Perform the specified operation
    if operation == "add":
        return a + b
    elif operation == "subtract":
        return a - b
    elif operation == "multiply":
        return a * b
    elif operation == "divide":
        if b == 0:
            raise ValueError("Division by zero is not allowed.")
        return a / b
    else:
        raise ValueError(f"Invalid operation: {operation}")
```

```python
agent= create_agent(
    model="openai:gpt-5-mini",
    tools=[real_number_calculator],
    system_prompt="You are a helpful assistant",
)
```

```python
result= agent.invoke("message"[{"rule": "user", "content": "What is 3 * 4"}])
```
